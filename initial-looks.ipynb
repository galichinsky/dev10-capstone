```python
import pandas as pd
print(pd.__version__)

df_QOL = pd.read_csv('Quality_of_Life.csv')
df_QOL.dropna(inplace=True)
df_QOL = df_QOL.apply(lambda s: s.str.replace("'", '', regex=False) if s.dtype == "object" else s)
# rename columns
df_QOL.columns = ['country', 'purchasing_power_val', 'purchasing_power_cat', 'safety_value', 'safety_cat', 'healthcare_value', 'healthcare_cat', 'climate_val', 'climate_cat', 'cost_of_living_val', 'cost_of_living_cat', 'property_price_income_val', 'property_price_income_cat', 'traffic_commute_time_val', 'traffic_commute_time_cat', 'pollution_val', 'pollution_cat', 'quality_of_life_val', 'quality_of_life_cat']

# Remove ':' from the 'quality_of_life_val' column
df_QOL['quality_of_life_val'] = df_QOL['quality_of_life_val'].str.replace(':', '', regex=False).astype(float)
df_QOL['property_price_income_val'] = df_QOL['property_price_income_val'].str.replace(',', '', regex=False).astype(float)
df_QOL.info()
df_QOL.head()
```

```python
df_WHR = pd.read_csv('World-happiness-report-2024.csv')
df_WHR.dropna(inplace=True)
df_WHR.columns = ['country', 'region', 'ladder_score', 'upperwhisker', 'lowerwhisker', 'log_gdp', 'social_support', 'healthy_life_expectancy', 'freedom', 'generosity', 'perceptions_of_corruption', 'dystopia_residual']
df_WHR.info()
df_WHR.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot distribution of Quality of Life values
plt.figure(figsize=(10, 6))
sns.histplot(df_QOL['quality_of_life_val'], kde=True, bins=30, color='blue')
plt.title('Distribution of Quality of Life Values')
plt.xlabel('Quality of Life Value')
plt.ylabel('Frequency')
plt.show()

# Plot distribution of Ladder Score (Happiness)
plt.figure(figsize=(10, 6))
sns.histplot(df_WHR['ladder_score'], kde=True, bins=30, color='green')
plt.title('Distribution of Ladder Scores (Happiness)')
plt.xlabel('Ladder Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# Select only numeric columns from the Quality of Life dataset
numeric_df_QOL = df_QOL.select_dtypes(include=['number'])

# Correlation matrix for Quality of Life dataset
print("\nCorrelation Matrix for Quality of Life Dataset:")
print(numeric_df_QOL.corr())

# Heatmap of correlations for Quality of Life dataset
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df_QOL.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap - Quality of Life Dataset')
plt.show()
```
